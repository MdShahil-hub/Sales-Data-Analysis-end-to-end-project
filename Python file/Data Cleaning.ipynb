```python
import pandas as pd
```

```python
df = pd.read_excel("sales_raw_500.xlsx")
```

```python
data = df.copy()
```

```python
data.info()
```

```python
data.head()
```

### Remove duplicate coloumns

```python
data.duplicated().sum()
```

```python
data.shape
```

```python
data = data.drop_duplicates()
```

```python
data.shape
```

### Handle missing data value

```python
data[data['Order_Date'].isnull()]
```

```python
data['Order_Date'].isnull().mean() * 100
```

```python
data['Order_Date'] = pd.to_datetime(data['Order_Date'], errors='coerce')
# Agar kisi value ko date mein convert nahi kar sakte, toh Pandas error dene ke bajaye usko NaT bana deta hai.
```

```python
data = data.dropna(subset=['Order_Date'])
```

```python
data.head()
```

### Add new coloums

```python
data['Profit'] =  data['Sales']-data['Cost']
data['Year'] = data['Order_Date'].dt.year
data['Month'] = data['Order_Date'].dt.month
```
